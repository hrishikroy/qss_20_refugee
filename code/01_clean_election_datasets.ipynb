```python
import pandas as pd
from pathlib import Path
import numpy as np 

# Helper: safely sum coalition columns, treating missing parties as 0
def sum_existing_cols(df, cols):
    return df.reindex(columns=cols, fill_value=0).sum(axis=1)

def row_sum_existing_cols(df, cols):
    existing_cols = [col for col in cols if col in df.columns]
    
    if len(existing_cols) == 0:
        return pd.Series(0, index=df.index)
    
    return df[existing_cols].sum(axis=1, skipna=True)

# Chamber of Deputies Elections by Commune in 2018

commune_results_2018 = pd.read_csv(
    "/Users/hrishikroy/Downloads/camera-20180304/Camera2018_livComune.txt",
    sep=";",
    encoding="latin1")

commune_results_2018_pivoted = (commune_results_2018
    .pivot_table(
        index=["CIRCOSCRIZIONE", "COMUNE", "ELETTORI",
            "ELETTORI_MASCHI", "VOTANTI", "VOTANTI_MASCHI",
            "SCHEDE_BIANCHE"], columns="LISTA",
        values="VOTI_LISTA",
        aggfunc="sum",
        fill_value=0).reset_index())

commune_results_2018_pivoted.columns.name = None

commune_results_2018_pivoted.to_csv("/Users/hrishikroy/R Studio/Summer Research/data/commune_results_2018_pivoted.csv",
    index=False)



# Chamber of Deputies Elections by Commune in 2022


commune_results_2022 = pd.read_csv("/Users/hrishikroy/Downloads/camera-20220925/Camera_Italia_LivComune.csv",
    sep=";",
    decimal=",")

commune_results_2022_pivoted = (commune_results_2022
    .pivot_table(index=["CIRC-REG", "COMUNE", "ELETTORITOT", "ELETTORIM",
            "VOTANTITOT", "VOTANTIM", "SKBIANCHE"],
        columns="DESCRLISTA",
        values="VOTILISTA",
        aggfunc="sum",
        fill_value=0).reset_index())

commune_results_2022_pivoted.to_csv("/Users/hrishikroy/R Studio/Summer Research/data/commune_results_election_2022.csv",
    index=False)
```
